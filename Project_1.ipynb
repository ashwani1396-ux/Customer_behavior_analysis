```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median())) # filling null values with median
```

```python
df.isnull().sum()  # Checking Null Values Again
```

```python
df.column = df.columns.str.lower()
df.column = df.columns.str.replace(' ','_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# create a column age_group
labels=['Young Adult', 'Adult', 'Middle-Aged', 'Senior']
df['age_group'] = pd.qcut(df['Age'], q=4, labels=labels)
```

```python
df[['Age', 'age_group']].head(10)
```

```python
# create column purchase _frequency_days
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['Frequency of Purchases'].map(frequency_mapping)
```

```python
df['frequency_of_purchase'] = df['Frequency of Purchases']

df[['purchase_frequency_days', 'frequency_of_purchase']].head(10)
```

```python
df = df.rename(columns={
    'Discount Applied': 'discount_applied',
    'Promo Code Used': 'promo_code_used'
})
```

```python
df[['discount_applied','promo_code_used']].head(10)
```

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
df = df.drop('promo_code_used', axis=1)
```
